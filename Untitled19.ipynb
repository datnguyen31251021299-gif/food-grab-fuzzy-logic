```python
import os
import json
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau, ModelCheckpoint
from sklearn.metrics import classification_report, confusion_matrix
```

```python
tf.random.set_seed(99)
np.random.seed(99)
!cp /content/drive/MyDrive/dataset.zip /content/dataset.zip
!unzip -q /content/dataset.zip -d /content/local_dataset
DRIVE_DATASET_DIR = "/content/drive/MyDrive/canteen_project_dataset"
WORKSPACE_DIR = "/content/drive/MyDrive/canteen_ai_workspace"

MODEL_SAVE_PATH = os.path.join(WORKSPACE_DIR, "canteen_model.keras")
CLASS_LABELS_PATH = os.path.join(WORKSPACE_DIR, "class_names.json")

os.makedirs(WORKSPACE_DIR, exist_ok=True)

IMG_SHAPE = (224, 224)
BATCH_SIZE = 32

AUTOTUNE = tf.data.AUTOTUNE

if not os.path.exists(DRIVE_DATASET_DIR):
    raise ValueError(f"Không tìm thấy thư mục dataset tại: {DRIVE_DATASET_DIR}")

print("Đang tải dữ liệu từ Drive...")
```

```python
train_dataset = tf.keras.utils.image_dataset_from_directory(
    DRIVE_DATASET_DIR,
    validation_split=0.15,
    subset="training",
    seed=99,
    image_size=IMG_SHAPE,
    batch_size=BATCH_SIZE,
    label_mode="categorical"
)

val_dataset = tf.keras.utils.image_dataset_from_directory(
    DRIVE_DATASET_DIR,
    validation_split=0.15,
    subset="validation",
    seed=99,
    image_size=IMG_SHAPE,
    batch_size=BATCH_SIZE,
    label_mode="categorical"
)

class_names = train_dataset.class_names
num_classes = len(class_names)

with open(CLASS_LABELS_PATH, "w", encoding="utf-8") as f:
    json.dump(class_names, f, ensure_ascii=False, indent=4)
print(f"Đã lưu {num_classes} nhãn món ăn vào {CLASS_LABELS_PATH}")

train_dataset = train_dataset.shuffle(1000).prefetch(buffer_size=AUTOTUNE)
val_dataset = val_dataset.prefetch(buffer_size=AUTOTUNE)

def build_canteen_model(num_classes):
    base_cnn = tf.keras.applications.EfficientNetB0(
        input_shape=IMG_SHAPE + (3,),
        include_top=False,
        weights="imagenet"
    )

    base_cnn.trainable = False

    data_aug = tf.keras.Sequential([
        layers.RandomFlip("horizontal_and_vertical"),
        layers.RandomRotation(0.15),
        layers.RandomZoom(0.15),
        layers.RandomBrightness(factor=0.2)
    ], name="augmentation_block")


    inputs = layers.Input(shape=IMG_SHAPE + (3,))
    x = data_aug(inputs)
    x = base_cnn(x, training=False)

    x = layers.GlobalAveragePooling2D()(x)
    x = layers.Dense(512, activation='swish')(x)
    x = layers.Dropout(0.45)(x)
    outputs = layers.Dense(num_classes, activation="softmax")(x)

    model = models.Model(inputs, outputs)
    return model, base_cnn

model, base_cnn = build_canteen_model(num_classes)
class_weights = {}
total_images = 0
counts = []

for i, cls in enumerate(class_names):
    cls_path = os.path.join(DRIVE_DATASET_DIR, cls)
    if os.path.isdir(cls_path):
        num_imgs = len([f for f in os.listdir(cls_path) if f.lower().endswith(('.png', '.jpg', '.jpeg', '.webp'))])
        counts.append(num_imgs)
        total_images += num_imgs
    else:
        counts.append(0)

for i, count in enumerate(counts):
    if count > 0:
        class_weights[i] = total_images / (num_classes * count)
    else:
        class_weights[i] = 1.0

print("Trọng số (Class Weights) đã tính:", class_weights)
callbacks_list = [
    EarlyStopping(monitor="val_loss", patience=5, restore_best_weights=True, verbose=1),
    ReduceLROnPlateau(monitor="val_loss", factor=0.5, patience=2, min_lr=1e-6, verbose=1),
    ModelCheckpoint(MODEL_SAVE_PATH, monitor="val_accuracy", save_best_only=True, verbose=1)
]
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-3),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

history_phase1 = model.fit(
    train_dataset,
    validation_data=val_dataset,
    epochs=15,
    callbacks=callbacks_list
)
base_cnn.trainable = True
for layer in base_cnn.layers[:-30]:
    layer.trainable = False

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-4),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

history_phase2 = model.fit(
    train_dataset,
    validation_data=val_dataset,
    class_weight=class_weights,
    epochs=15,
    callbacks=callbacks_list
)

model.save(MODEL_SAVE_PATH)
print(f"Hoàn tất! Model đã được lưu tại: {MODEL_SAVE_PATH}")
```

```python
print("\n--- ĐÁNH GIÁ TRÊN TẬP VALIDATION ---")
y_actual = []
y_predicted = []

for imgs, lbls in val_dataset:
    preds = model.predict(imgs, verbose=0)
    y_predicted.extend(np.argmax(preds, axis=1))
    y_actual.extend(np.argmax(lbls.numpy(), axis=1))

print("\nClassification Report:")
print(classification_report(y_actual, y_predicted, target_names=class_names, zero_division=0))

cm = confusion_matrix(y_actual, y_predicted)
plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", xticklabels=class_names, yticklabels=class_names)
plt.title("Confusion Matrix - Canteen Project", fontweight="bold", fontsize=14)
plt.xlabel("Mô hình Dự đoán", fontsize=12)
plt.ylabel("Nhãn Thực tế", fontsize=12)
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```
