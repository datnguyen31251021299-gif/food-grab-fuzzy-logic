```python
# ==============================================================================
# BƯỚC 1: KHỞI TẠO MÔI TRƯỜNG & DỌN DẸP HỆ THỐNG GỐC
# ==============================================================================
from google.colab import drive
import os
import shutil

print("🧹 Đang tiến hành kiểm tra và giải phóng cổng kết nối cũ...")
try:
    drive.flush_and_unmount()
except:
    pass

if os.path.exists('/content/drive') and not os.listdir('/content/drive'):
    try:
        shutil.rmtree('/content/drive')
    except:
        pass

# Cài đặt các thư viện cần thiết (Bỏ nest_asyncio để tránh xung đột uvicorn)
!apt-get install -y fonts-dejavu-core -q
!pip install gradio opencv-python numpy tensorflow reportlab pillow -q

import cv2
import json
import logging
import gc
import datetime
import numpy as np
import tensorflow as tf
from PIL import Image
import gradio as gr
from collections import Counter

# Tiến hành Mount Drive an toàn
try:
    print("📂 Đang kết nối tới Google Drive...")
    drive.mount('/content/drive', force_remount=True)
    print("✅ Kết nối Drive thành công.")
except ValueError as e:
    if "Mountpoint must not already contain files" in str(e):
        print("ℹ️ Drive đã được kết nối từ trước.")
    else:
        raise e

# Thư viện ReportLab xuất hóa đơn PDF
from reportlab.lib.pagesizes import letter
from reportlab.platypus import SimpleDocTemplate, Paragraph, Table, TableStyle
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
from reportlab.lib import colors
from reportlab.pdfbase import pdfmetrics
from reportlab.pdfbase.ttfonts import TTFont

# ==============================================================================
# SECTION 1: SYSTEM LOGGING & MONITORING PIPELINE
# ==============================================================================
logging.basicConfig(
    level=logging.INFO,
    format='%(asctime)s [%(levelname)s] %(name)s: %(message)s',
    handlers=[
        logging.StreamHandler(),
        logging.FileHandler("canteen_pos_system.log", encoding="utf-8")
    ]
)
logger = logging.getLogger("SmartCanteenPOS")

# ==============================================================================
# SECTION 2: ĐƯỜNG DẪN TRỰC TIẾP TRÊN GOOGLE DRIVE
# ==============================================================================
WORKSPACE_DIR = "/content/drive/MyDrive/canteen_ai_workspace"
MODEL_PATH = '/content/drive/MyDrive/canteen_ai_workspace/canteen_model.keras'
CLASS_LABELS_PATH = '/content/drive/MyDrive/canteen_ai_workspace/class_names.json'
FEEDBACK_DIR = '/content/drive/MyDrive/canteen_ai_workspace/ai_feedback_dataset'

QR_PATH = '/content/drive/MyDrive/canteen_ai_workspace/QR.jpg'
PDF_OUTPUT_PATH = os.path.join(WORKSPACE_DIR, "hoa_don_khay_com.pdf")
IMG_SHAPE = (224, 224)

os.makedirs(FEEDBACK_DIR, exist_ok=True)

# Cấu hình Fonts hiển thị tiếng Việt trên PDF ReportLab
FONT_PATH = "/usr/share/fonts/truetype/dejavu/DejaVuSans.ttf"
FONT_BOLD_PATH = "/usr/share/fonts/truetype/dejavu/DejaVuSans-Bold.ttf"

if os.path.exists(FONT_PATH) and os.path.exists(FONT_BOLD_PATH):
    try:
        pdfmetrics.registerFont(TTFont('DejaVuSans', FONT_PATH))
        pdfmetrics.registerFont(TTFont('DejaVuSans-Bold', FONT_BOLD_PATH))
        FONT_NAME = "DejaVuSans"
        FONT_BOLD_NAME = "DejaVuSans-Bold"
    except Exception as e:
        FONT_NAME = "Helvetica"
        FONT_BOLD_NAME = "Helvetica-Bold"
else:
    FONT_NAME = "Helvetica"
    FONT_BOLD_NAME = "Helvetica-Bold"

FULL_CLASS_NAMES = [
    "ca", "canh chua", "canh chua co ca", "com",
    "dau hu", "rau", "suon", "thit kho", "thit kho trung", "trung"
]

# Tải mô hình mạng nơ-ron
try:
    if os.path.exists(MODEL_PATH):
        model = tf.keras.models.load_model(MODEL_PATH)
        logger.info("✅ Đã tải mạng CNN Core Model thành công.")
    else:
        logger.warning("⚠️ Không tìm thấy file mô hình Keras. Chuyển sang Engine chạy mô phỏng.")
        model = None

    if os.path.exists(CLASS_LABELS_PATH):
        with open(CLASS_LABELS_PATH, "r", encoding="utf-8") as f:
            class_names = json.load(f)
    else:
        class_names = FULL_CLASS_NAMES
except Exception as e:
    logger.critical(f"💥 Lỗi khởi tạo Model Pipeline: {str(e)}")
    model = None
    class_names = FULL_CLASS_NAMES

PRICE_MAP = {
    "ca": {"name": "Cá hú kho", "price": 30000},
    "canh chua": {"name": "Canh chua không cá", "price": 10000},
    "canh chua co ca": {"name": "Canh chua có cá", "price": 25000},
    "com": {"name": "Cơm trắng", "price": 10000},
    "dau hu": {"name": "Đậu hũ sốt cà", "price": 25000},
    "rau": {"name": "Rau xào", "price": 10000},
    "suon": {"name": "Sườn nướng", "price": 30000},
    "thit kho": {"name": "Thịt kho", "price": 25000},
    "thit kho trung": {"name": "Thịt kho trứng", "price": 30000},
    "trung": {"name": "Trứng chiên", "price": 25000}
}

def get_dish_info(label_key):
    clean_key = label_key.strip().lower()
    if clean_key in PRICE_MAP:
        return PRICE_MAP[clean_key]["name"], PRICE_MAP[clean_key]["price"]
    return label_key, 0

LAYOUT_2_TOP_3_BOTTOM = [
    {"slot": "Ô 1 (Trên trái)", "box": [0.05, 0.05, 0.45, 0.45]},
    {"slot": "Ô 2 (Trên phải)", "box": [0.05, 0.55, 0.45, 0.95]},
    {"slot": "Ô 3 (Dưới trái)", "box": [0.55, 0.05, 0.95, 0.35]},
    {"slot": "Ô 4 (Dưới giữa)", "box": [0.55, 0.40, 0.95, 0.65]},
    {"slot": "Ô 5 (Dưới phải)", "box": [0.55, 0.70, 0.95, 0.95]}
]

LAYOUT_3_TOP_2_BOTTOM = [
    {"slot": "Ô 1 (Trên trái)", "box": [0.05, 0.05, 0.45, 0.35]},
    {"slot": "Ô 2 (Trên giữa)", "box": [0.05, 0.35, 0.45, 0.65]},
    {"slot": "Ô 3 (Trên phải)", "box": [0.05, 0.65, 0.45, 0.95]},
    {"slot": "Ô 4 (Dưới trái)", "box": [0.55, 0.15, 0.95, 0.45]},
    {"slot": "Ô 5 (Dưới phải)", "box": [0.55, 0.55, 0.95, 0.85]}
]

# ==============================================================================
# SECTION 3: ADVANCED INDUSTRIAL COMPUTER VISION ENGINE
# ==============================================================================
class IndustrialVisionEngine:
    @staticmethod
    def auto_white_balance(img):
        b, g, r = cv2.split(img)
        b_avg, g_avg, r_avg = np.mean(b), np.mean(g), np.mean(r)
        k = (b_avg + g_avg + r_avg) / 3
        b = np.clip(b * (k / (b_avg + 1e-6)), 0, 255).astype(np.uint8)
        g = np.clip(g * (k / (g_avg + 1e-6)), 0, 255).astype(np.uint8)
        r = np.clip(r * (k / (g_avg + 1e-6)), 0, 255).astype(np.uint8)
        return cv2.merge((b, g, r))

    @staticmethod
    def advanced_enhance_crop(crop_img):
        if crop_img is None or crop_img.size == 0: return crop_img
        try:
            gamma = 1.2
            invGamma = 1.0 / gamma
            table = np.array([((i / 255.0) ** invGamma) * 255 for i in np.arange(0, 256)]).astype("uint8")
            enhanced_bgr = cv2.LUT(crop_img, table)
            smoothed = cv2.bilateralFilter(enhanced_bgr, d=5, sigmaColor=50, sigmaSpace=50)
            return smoothed
        except Exception as e:
            return crop_img

    @staticmethod
    def deskew_image(img):
        try:
            gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
            blur = cv2.GaussianBlur(gray, (9, 9), 0)
            edges = cv2.Canny(blur, 50, 150, apertureSize=3)
            lines = cv2.HoughLinesP(edges, 1, np.pi/180, 100, minLineLength=200, maxLineGap=10)
            if lines is not None:
                angles = [np.arctan2(y2-y1, x2-x1)*180/np.pi for line in lines for x1,y1,x2,y2 in line]
                angles = [a for a in angles if -45 < a < 45]
                if angles:
                    median_angle = np.median(angles)
                    if abs(median_angle) > 0.5:
                        (h, w) = img.shape[:2]
                        M = cv2.getRotationMatrix2D((w//2, h//2), median_angle, 1.0)
                        return cv2.warpAffine(img, M, (w, h), flags=cv2.INTER_CUBIC, borderMode=cv2.BORDER_REPLICATE)
            return img
        except Exception as e:
            return img

# ==============================================================================
# SECTION 4: DEEP LEARNING (CNN) INFERENCE & HEURISTIC ENGINE
# ==============================================================================
class AIInferenceEngine:
    @staticmethod
    def preprocess_tensor_batch(crop_list):
        tensor_batch = []
        for idx, crop in enumerate(crop_list):
            if crop is None or crop.size == 0:
                crop = np.zeros((IMG_SHAPE[0], IMG_SHAPE[1], 3), dtype=np.uint8)

            crop_rgb = cv2.cvtColor(crop, cv2.COLOR_BGR2RGB)
            img_resized = cv2.resize(crop_rgb, IMG_SHAPE, interpolation=cv2.INTER_AREA)
            img_float = img_resized.astype(np.float32)
            tensor_batch.append(img_float)

        np_batch = np.array(tensor_batch, dtype=np.float32)
        return tf.convert_to_tensor(np_batch, dtype=tf.float32)

    @staticmethod
    def apply_smart_heuristic_filters(preds_batch, class_names, conf_threshold, smart_filter_active):
        output_labels = []
        confidences = []
        status_meta = ["Normal"] * 5

        for dist in preds_batch:
            idx = np.argmax(dist)
            output_labels.append(class_names[idx])
            confidences.append(float(dist[idx]))

        if not smart_filter_active:
            return output_labels, status_meta, confidences

        for idx, dist in enumerate(preds_batch):
            top_2_indices = np.argsort(dist)[-2:][::-1]
            first_idx, second_idx = top_2_indices[0], top_2_indices[1]

            first_label = class_names[first_idx]
            second_label = class_names[second_idx]
            margin = dist[first_idx] - dist[second_idx]

            if first_label in ["thit kho", "suon"] and second_label == "dau hu" and margin < 0.25:
                output_labels[idx] = "dau hu"
                confidences[idx] = float(dist[second_idx])
                status_meta[idx] = "Tự động sửa (Đậu hũ)"
                continue

            if first_label == "suon" and second_label in ["thit kho trung", "thit kho"] and margin < 0.20:
                output_labels[idx] = second_label
                confidences[idx] = float(dist[second_idx])
                status_meta[idx] = "Tự động sửa (Thịt kho)"
                continue

        counts = Counter(output_labels)
        if counts["suon"] >= 1 and counts["thit kho"] >= 1:
            for idx, lbl in enumerate(output_labels):
                if lbl in ["suon", "thit kho"] and confidences[idx] < 0.80 and "Tự động sửa" not in status_meta[idx]:
                    status_meta[idx] = "Nghi ngờ trùng món mặn"

        for idx, score in enumerate(confidences):
            if score < conf_threshold and "Tự động sửa" not in status_meta[idx]:
                status_meta[idx] = "Độ tin cậy yếu"

        return output_labels, status_meta, confidences

# ==============================================================================
# SECTION 5: UNICODE COMPLIANT PDF INVOICE GENERATOR
# ==============================================================================
def generate_pdf_invoice(items, total_amount):
    try:
        doc = SimpleDocTemplate(PDF_OUTPUT_PATH, pagesize=letter, rightMargin=40, leftMargin=40, topMargin=40, bottomMargin=40)
        story = []
        styles = getSampleStyleSheet()

        title_style = ParagraphStyle('InvoiceTitle', parent=styles['Heading1'], fontName=FONT_BOLD_NAME, fontSize=20, leading=24, textColor=colors.HexColor('#14532d'), alignment=1, spaceAfter=15)
        meta_style = ParagraphStyle('InvoiceMeta', parent=styles['Normal'], fontName=FONT_NAME, fontSize=10, leading=14, textColor=colors.HexColor('#475569'), spaceAfter=15)
        cell_style = ParagraphStyle('InvoiceCell', parent=styles['Normal'], fontName=FONT_NAME, fontSize=10, leading=14, textColor=colors.HexColor('#1e293b'))
        cell_bold = ParagraphStyle('InvoiceCellBold', parent=styles['Normal'], fontName=FONT_BOLD_NAME, fontSize=10, leading=14, textColor=colors.HexColor('#0f172a'))

        story.append(Paragraph("HÓA ĐƠN THANH TOÁN - AI SMART CANTEEN", title_style))
        story.append(Paragraph(f"<b>Thời gian:</b> {datetime.datetime.now().strftime('%d/%m/%Y %H:%M:%S')}<br/><b>Mã GD:</b> TXN-{int(datetime.datetime.now().timestamp())}", meta_style))

        table_data = [[Paragraph("<b>Vị trí ô khay</b>", cell_bold), Paragraph("<b>Món ăn nhận diện</b>", cell_bold), Paragraph("<b>Đơn giá</b>", cell_bold)]]
        for item in items:
            table_data.append([Paragraph(str(item['slot']), cell_style), Paragraph(str(item['name']), cell_style), Paragraph(f"{item['price']:,} đ", cell_style)])

        table_data.append([Paragraph("<b>TỔNG THÀNH TIỀN</b>", cell_bold), "", Paragraph(f"<b>{total_amount:,} đ</b>", cell_bold)])

        invoice_table = Table(table_data, colWidths=[150, 230, 150])
        invoice_table.setStyle(TableStyle([
            ('BACKGROUND', (0,0), (-1,0), colors.HexColor('#f1f5f9')),
            ('ALIGN', (0,0), (-1,-1), 'LEFT'),
            ('VALIGN', (0,0), (-1,-1), 'MIDDLE'),
            ('LINEBELOW', (0,0), (-1,-2), 0.5, colors.HexColor('#cbd5e1')),
            ('LINEBELOW', (0,-1), (-1,-1), 1.5, colors.HexColor('#14532d')),
            ('SPAN', (0, -1), (1, -1)),
            ('PADDING', (0,0), (-1,-1), 10),
        ]))
        story.append(invoice_table)
        doc.build(story)
        return PDF_OUTPUT_PATH
    except Exception as e:
        logger.error(f"❌ Lỗi xuất PDF: {str(e)}")
        return None

# ==============================================================================
# SECTION 6-9: CORE PROCESSOR PIPELINE
# ==============================================================================
def process_food_scanner(cam_img, file_img, conf_threshold, border_ratio, smart_filter_active, deskew_active, layout_choice, session_state):
    input_img = file_img if file_img is not None else cam_img
    if input_img is None:
        return [None, "<div style='color:#ef4444; font-weight:bold; padding: 15px; border-radius: 8px; background: #fee2e2;'>❌ Lỗi: Hãy cung cấp hình ảnh!</div>", None, gr.update(value=None), [], session_state] + ["com"] * 5

    active_layout = LAYOUT_3_TOP_2_BOTTOM if layout_choice == "3 Ô Trên - 2 Ô Dưới" else LAYOUT_2_TOP_3_BOTTOM
    session_state["layout_choice"] = layout_choice

    print("🚀 Phân tích Thị Giác Máy Tính...")
    work_img = np.array(input_img.copy(), dtype=np.uint8)
    work_img = cv2.cvtColor(work_img, cv2.COLOR_RGB2BGR)
    work_img = IndustrialVisionEngine.auto_white_balance(work_img)
    if deskew_active:
        work_img = IndustrialVisionEngine.deskew_image(work_img)

    h, w, _ = work_img.shape
    debug_img = work_img.copy()

    crop_list = []
    detected_items = []
    total_bill = 0
    table_report_data = []

    for i, cfg in enumerate(active_layout):
        ymin, xmin, ymax, xmax = cfg["box"]
        x1, y1 = int(xmin * w), int(ymin * h)
        x2, y2 = int(xmax * w), int(ymax * h)

        bw, bh = x2 - x1, y2 - y1
        x1_adj, x2_adj = max(0, min(int(x1 + bw * border_ratio), w-1)), max(0, min(int(x2 - bw * border_ratio), w))
        y1_adj, y2_adj = max(0, min(int(y1 + bh * border_ratio), h-1)), max(0, min(int(y2 - bh * border_ratio), h))

        crop = work_img[y1_adj:y2_adj, x1_adj:x2_adj]
        if crop.size == 0:
            crop = np.zeros((100, 100, 3), dtype=np.uint8)

        crop_enhanced = IndustrialVisionEngine.advanced_enhance_crop(crop)
        crop_list.append(crop_enhanced)
        session_state["crops"][i] = cv2.cvtColor(crop_enhanced, cv2.COLOR_BGR2RGB)

        cv2.rectangle(debug_img, (x1, y1), (x2, y2), (16, 185, 129), 4)
        cv2.putText(debug_img, f"Slot {i+1}", (x1 + 10, y1 + 35), cv2.FONT_HERSHEY_SIMPLEX, 1.1, (16, 185, 129), 3)

    batch_tensors = AIInferenceEngine.preprocess_tensor_batch(crop_list)

    if model is not None:
        preds_tensor = model(batch_tensors, training=False)
        preds_batch = preds_tensor.numpy()
    else:
        preds_batch = np.zeros((5, len(class_names)))
        mock_labels = ["com", "rau", "dau hu", "trung", "thit kho trung"]
        for idx, lbl in enumerate(mock_labels):
            class_idx = class_names.index(lbl)
            preds_batch[idx][class_idx] = 0.95

    final_labels, status_meta, confidences = AIInferenceEngine.apply_smart_heuristic_filters(
        preds_batch, class_names, conf_threshold, smart_filter_active
    )

    for i, label in enumerate(final_labels):
        slot_name = active_layout[i]["slot"]
        dish_name, price = get_dish_info(label)
        total_bill += price
        session_state["predictions"][i] = label

        conf_score = confidences[i] * 100
        if "Tự động sửa" in status_meta[i]: badge = f"🟢 {status_meta[i]} ({conf_score:.1f}%)"
        elif "Nghi ngờ" in status_meta[i] or "yếu" in status_meta[i]: badge = f"🔴 {status_meta[i]} ({conf_score:.1f}%)"
        elif conf_score >= 85: badge = f"🟢 Tin cậy cao ({conf_score:.1f}%)"
        else: badge = f"🟡 Nghi ngờ ({conf_score:.1f}%)"

        detected_items.append({"slot": slot_name, "name": dish_name, "price": price})
        table_report_data.append([slot_name, dish_name, f"{price:,} đ", badge])

    html_receipt = f"""
    <div style='background: linear-gradient(135deg, #f0fdf4 0%, #dcfce7 100%); border: 1px solid #bbf7d0; padding: 20px; border-radius: 12px; font-family: system-ui, sans-serif;'>
        <span style='color: #166534; font-size: 13px; font-weight: 700; text-transform: uppercase; letter-spacing: 0.05em; display: block; margin-bottom: 4px;'>💰 TỔNG THÀNH TIỀN</span>
        <h1 style='color: #15803d; font-size: 46px; font-weight: 900; margin: 0; font-variant-numeric: tabular-nums; display: flex; align-items: baseline;'>
            {total_bill:,} <span style='font-size: 22px; font-weight: 700; margin-left: 6px;'>đ</span>
        </h1>
    </div>
    """
    qr_img = Image.open(QR_PATH) if os.path.exists(QR_PATH) else Image.new("RGB", (300, 300), color="#16a34a")
    pdf_file = generate_pdf_invoice(detected_items, total_bill)
    debug_img_rgb = cv2.cvtColor(debug_img, cv2.COLOR_BGR2RGB)

    del work_img, debug_img, batch_tensors
    gc.collect()

    return [debug_img_rgb, html_receipt, qr_img, pdf_file, table_report_data, session_state] + final_labels


def update_manual_bill(d1, d2, d3, d4, d5, session_state):
    selected_labels = [d1, d2, d3, d4, d5]
    total_bill = 0
    detected_items = []
    table_report_data = []

    active_layout = LAYOUT_3_TOP_2_BOTTOM if session_state.get("layout_choice") == "3 Ô Trên - 2 Ô Dưới" else LAYOUT_2_TOP_3_BOTTOM

    for i, pred_label in enumerate(selected_labels):
        cfg = active_layout[i]
        dish_name, price = get_dish_info(pred_label)
        total_bill += price
        session_state["predictions"][i] = pred_label
        table_report_data.append([cfg["slot"], dish_name, f"{price:,} đ", "🔵 Thu ngân ghi đè"])
        detected_items.append({"slot": cfg["slot"], "name": dish_name, "price": price})

    html_receipt = f"""
    <div style='background: linear-gradient(135deg, #eff6ff 0%, #dbeafe 100%); border: 1px solid #bfdbfe; padding: 20px; border-radius: 12px; font-family: system-ui, sans-serif;'>
        <span style='color: #1e40af; font-size: 13px; font-weight: 700; text-transform: uppercase; letter-spacing: 0.05em; display: block; margin-bottom: 4px;'>💰 TỔNG TIỀN (ĐÃ ĐỀ NHÃN)</span>
        <h1 style='color: #1d4ed8; font-size: 46px; font-weight: 900; margin: 0; font-variant-numeric: tabular-nums; display: flex; align-items: baseline;'>
            {total_bill:,} <span style='font-size: 22px; font-weight: 700; margin-left: 6px;'>đ</span>
        </h1>
    </div>
    """
    pdf_file = generate_pdf_invoice(detected_items, total_bill)
    return html_receipt, pdf_file, table_report_data, session_state

def submit_feedback_dataset(d1, d2, d3, d4, d5, session_state):
    user_labels = [d1, d2, d3, d4, d5]
    saved_count = 0
    timestamp = datetime.datetime.now().strftime("%Y%m%d_%H%M%S")

    for i, crop_img in enumerate(session_state["crops"]):
        if crop_img is None: continue
        if user_labels[i] != session_state["predictions"][i]:
            try:
                class_dir = os.path.join(FEEDBACK_DIR, user_labels[i])
                os.makedirs(class_dir, exist_ok=True)
                filepath = os.path.join(class_dir, f"auto_err_{user_labels[i]}_{timestamp}_slot{i+1}.jpg")
                cv2.imwrite(filepath, cv2.cvtColor(crop_img, cv2.COLOR_RGB2BGR))
                saved_count += 1
            except Exception as e:
                logger.error(f"❌ Lỗi Active Learning: {str(e)}")

    if saved_count > 0:
        return f"<div style='color:#15803d; font-weight:700; padding: 12px; background: #dcfce7; border-radius: 8px; border: 1px solid #bbf7d0;'>🚀 Đồng bộ thành công: Đã lưu {saved_count} vùng ảnh sửa đổi!</div>"
    return "<div style='color:#1d4ed8; font-weight:700; padding: 12px; background: #dbeafe; border-radius: 8px; border: 1px solid #bfdbfe;'>💡 Trạng thái: Nhãn hiện tại trùng khớp. Không ghi nhận biến động.</div>"

def get_selected_crop_viewer(evt: gr.SelectData, session_state):
    row_idx = evt.index[0]
    if 0 <= row_idx < len(session_state["crops"]) and session_state["crops"][row_idx] is not None:
        return session_state["crops"][row_idx]
    return None

# ==============================================================================
# SECTION 10: INDUSTRIAL STYLED APPARATUS VIEW LAYOUT GENERATOR
# ==============================================================================
custom_css = """
body { background-color: #f1f5f9; font-family: 'Inter', system-ui, -apple-system, sans-serif; color: #1e293b; }
.gradio-container { max-width: 1440px !important; padding: 24px !important; margin: 0 auto !important; }

/* Header Cao cấp */
.app-header { background: linear-gradient(135deg, #064e3b 0%, #022c22 100%); padding: 30px; border-radius: 18px; color: white; text-align: center; margin-bottom: 24px; box-shadow: 0 10px 25px -5px rgba(2, 44, 34, 0.12); }

/* Thẻ khối góc bo tròn hiện đại */
.section-card { background: #ffffff !important; border-radius: 16px !important; border: 1px solid #e2e8f0 !important; padding: 20px !important; margin-bottom: 20px !important; box-shadow: 0 4px 6px -1px rgba(0, 0, 0, 0.02) !important; transition: box-shadow 0.2s ease; }
.section-card:hover { box-shadow: 0 10px 15px -3px rgba(0, 0, 0, 0.04) !important; }

/* Thiết kế Nút Quét Lớn siêu mượt */
.green-btn { background: linear-gradient(135deg, #10b981 0%, #059669 100%) !important; color: white !important; font-size: 18px !important; font-weight: 700 !important; height: 58px !important; border-radius: 12px !important; border: none !important; cursor: pointer; width: 100%; margin: 16px 0 !important; box-shadow: 0 4px 14px 0 rgba(16, 185, 129, 0.3) !important; transition: all 0.2s ease-in-out !important; }
.green-btn:hover { transform: translateY(-2px) !important; box-shadow: 0 6px 20px 0 rgba(16, 185, 129, 0.4) !important; filter: brightness(1.04); }
.green-btn:active { transform: translateY(0px) !important; }

/* Nút phụ chỉnh sửa & đồng bộ */
.grey-btn { background: linear-gradient(135deg, #f8fafc 0%, #e2e8f0 100%) !important; color: #334155 !important; border: 1px solid #cbd5e1 !important; font-weight: 600 !important; border-radius: 10px !important; padding: 12px 20px !important; transition: all 0.2s !important; }
.grey-btn:hover { background: #e2e8f0 !important; color: #0f172a !important; }
.orange-btn { background: linear-gradient(135deg, #f97316 0%, #ea580c 100%) !important; color: white !important; font-weight: 600 !important; border-radius: 10px !important; border: none !important; padding: 12px 20px !important; box-shadow: 0 4px 10px 0 rgba(249, 115, 22, 0.15) !important; transition: all 0.2s !important; }
.orange-btn:hover { transform: translateY(-1px) !important; box-shadow: 0 6px 14px 0 rgba(249, 115, 22, 0.25) !important; }

/* Tinh chỉnh Dataframe và Tabs */
.nav-tab-custom { font-weight: 700 !important; font-size: 15px !important; }
.gradio-dataframe table { border-radius: 8px !important; overflow: hidden !important; }
footer { display: none !important; visibility: hidden !important; }
"""

# Truyền trực tiếp css vào Blocks theo chuẩn cấu trúc mới để fix deprecation warning
with gr.Blocks(title="SmartCanteen Industrial POS", css=custom_css) as demo:
    pos_state = gr.State({"crops": [None]*5, "predictions": ["com"]*5, "layout_choice": "2 Ô Trên - 3 Ô Dưới"})

    gr.HTML("""
    <div class='app-header'>
        <h1 style='margin: 0; font-size: 34px; font-weight: 800; letter-spacing: -0.5px;color: white;'>🥦 SmartCanteen Industrial POS System</h1>
        <p style='margin: 6px 0 0 0; font-size: 15px; opacity: 0.9; font-weight: 500;color: white;'>Hệ thống Quét Khay Ăn AI Đa Nền Tảng - Phân Phối Cấp Độ Doanh Nghiệp</p>
    </div>
    """)

    with gr.Tabs():
        with gr.TabItem("🏪 MÀN HÌNH BÀN HÀNG (POS)", elem_classes="nav-tab-custom"):
            with gr.Row(equal_height=False):
                with gr.Column(scale=6):
                    with gr.Column(elem_classes="section-card"):
                        with gr.Tabs():
                            with gr.TabItem("📸 CAMERA PHẦN CỨNG USB"):
                                cam_input = gr.Image(sources=["webcam"], type="numpy", label="Live Stream", height=400)
                            with gr.TabItem("💻 CAMERA TRÌNH DUYỆT / FILE UPLOAD"):
                                file_input = gr.Image(sources=["upload"], type="numpy", label="Static File Input", height=400)

                    with gr.Accordion("⚙️ Cấu hình bộ lọc màu AI & Cân biên nâng cao", open=True):
                        layout_dropdown = gr.Dropdown(choices=["2 Ô Trên - 3 Ô Dưới", "3 Ô Trên - 2 Ô Dưới"], value="2 Ô Trên - 3 Ô Dưới", label="📐 Cấu hình sơ đồ khay ăn")
                        with gr.Row():
                            sl_conf = gr.Slider(minimum=0.1, maximum=0.95, value=0.25, step=0.05, label="Ngưỡng tin cậy (Confidence)")
                            sl_border = gr.Slider(minimum=0.0, maximum=0.25, value=0.08, step=0.02, label="Tỉ lệ xén biên (Crop Margin)")
                        with gr.Row():
                            chk_filter = gr.Checkbox(value=True, label="Kích hoạt logic bộ lọc Heuristic")
                            chk_deskew = gr.Checkbox(value=True, label="Tự động cân phẳng khay (Auto-Deskew)")

                with gr.Column(scale=3):
                    with gr.Column(elem_classes="section-card"):
                        gr.HTML("<b style='color:#475569; font-size:13px; text-transform: uppercase; letter-spacing:0.5px; display:block; margin-bottom:10px;'>👁️ Thị giác máy tính phân rã</b>")
                        vision_output = gr.Image(label="Vision View", show_label=False, height=350)

                with gr.Column(scale=3):
                    with gr.Column(elem_classes="section-card"):
                        bill_html_output = gr.HTML("<div style='color:#64748b; padding: 25px 0; text-align: center; font-weight: 500;'>Hệ thống ở trạng thái sẵn sàng...</div>")
                    with gr.Column(elem_classes="section-card"):
                        gr.HTML("<b style='color:#475569; font-size:13px; text-transform: uppercase; letter-spacing:0.5px; display:block; margin-bottom:10px;'>📲 Thanh toán & Biên lai</b>")
                        qr_output = gr.Image(label="Mã QR Pay", show_label=False, height=160)
                        pdf_output = gr.File(label="Tải PDF")

            scan_btn = gr.Button("⚡ KÍCH HOẠT AI QUÉT KHAY & TÍNH TIỀN TỰ ĐỘNG", elem_classes="green-btn")

            with gr.Column(elem_classes="section-card"):
                gr.HTML("<b style='color:#1e293b; font-size:16px; display:block; margin-bottom:12px;'>📋 Chi tiết bóc tách phân tích & Điều chỉnh sai lệch</b>")
                with gr.Row():
                    with gr.Column(scale=7):
                        table_report = gr.Dataframe(headers=["Vị trí ô khay", "Món ăn nhận diện", "Đơn giá", "Trạng thái tin cậy AI"], row_count=5, col_count=4, interactive=False)
                    with gr.Column(scale=5):
                        crop_viewer = gr.Image(label="Focused Crop", show_label=False, height=165)

                gr.HTML("<div style='margin-top: 15px; margin-bottom: 5px; color: #475569; font-size: 13px; font-weight: 600;'>🛠️ BẢNG ĐIỀU CHỈNH NHANH NHÃN MÓN ĂN KHI AI NHẬN DIỆN CHƯA CHUẨN:</div>")
                with gr.Row():
                    d1 = gr.Dropdown(choices=class_names, label="Ô 1 (Trên trái)", value="com")
                    d2 = gr.Dropdown(choices=class_names, label="Ô 2 (Trên phải)", value="com")
                    d3 = gr.Dropdown(choices=class_names, label="Ô 3 (Dưới trái)", value="com")
                    d4 = gr.Dropdown(choices=class_names, label="Ô 4 (Dưới giữa)", value="com")
                    d5 = gr.Dropdown(choices=class_names, label="Ô 5 (Dưới phải)", value="com")

                with gr.Row():
                    update_btn = gr.Button("🔄 Cập nhật biên lai & Tính lại tiền", elem_classes="grey-btn")
                    feedback_btn = gr.Button("💾 Đóng góp ảnh sai sót vào Dataset (Active Learning)", elem_classes="orange-btn")

                status_html = gr.HTML("<div style='margin-top: 10px;'></div>")

        with gr.TabItem("📊 BÁO CÁO & THỐNG KÊ", elem_classes="nav-tab-custom"):
            with gr.Column(elem_classes="section-card"):
                gr.HTML("<h3 style='color:#064e3b; margin-top:0;'>📈 Biểu đồ phân tích doanh thu</h3>")
                gr.Markdown("Module giám sát hiệu năng đang được phát triển...")

    dropdowns = [d1, d2, d3, d4, d5]

    scan_btn.click(
        fn=process_food_scanner,
        inputs=[cam_input, file_input, sl_conf, sl_border, chk_filter, chk_deskew, layout_dropdown, pos_state],
        outputs=[vision_output, bill_html_output, qr_output, pdf_output, table_report, pos_state] + dropdowns
    )

    update_btn.click(
        fn=update_manual_bill,
        inputs=dropdowns + [pos_state],
        outputs=[bill_html_output, pdf_output, table_report, pos_state]
    )

    feedback_btn.click(
        fn=submit_feedback_dataset,
        inputs=dropdowns + [pos_state],
        outputs=status_html
    )

    table_report.select(
        fn=get_selected_crop_viewer,
        inputs=[pos_state],
        outputs=crop_viewer
    )

# ==============================================================================
# KHỞI CHẠY KHÔNG KHÓA LUỒNG (NON-BLOCKING LAUNCH CHO COLAB)
# ==============================================================================
if __name__ == "__main__":
    # prevent_thread_lock=True giúp ngắt chặn luồng, sửa lỗi đóng băng giao diện uvicorn
    demo.launch(share=True, prevent_thread_lock=True, debug=False)
```
