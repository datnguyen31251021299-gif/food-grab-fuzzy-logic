```python
import tensorflow as tf
from tensorflow.keras import layers, models
import matplotlib.pyplot as plt
import numpy as np


mnist = tf.keras.datasets.mnist
(x_train, y_train), (x_test, y_test) = mnist.load_data()


x_train, x_test = x_train / 255.0, x_test / 255.0
```

```python
def train_model(num_hidden_layers=1, learning_rate=0.001):
    model = models.Sequential()
    model.add(layers.Flatten(input_shape=(28, 28)))


    for _ in range(num_hidden_layers):
        model.add(layers.Dense(128, activation='relu'))

    model.add(layers.Dense(10, activation='softmax'))


    optimizer = tf.keras.optimizers.Adam(learning_rate=learning_rate)

    model.compile(optimizer=optimizer,
                  loss='sparse_categorical_crossentropy',
                  metrics=['accuracy'])

    history = model.fit(x_train, y_train, epochs=5, validation_split=0.2, verbose=0)
    return history
```

```python
h1 = train_model(num_hidden_layers=1)
h2 = train_model(num_hidden_layers=2)
h3 = train_model(num_hidden_layers=3)
h4 = train_model(num_hidden_layers=4)


plt.figure(figsize=(10, 6))

plt.plot(h1.history['val_accuracy'], label='1 Hidden Layer')
plt.plot(h2.history['val_accuracy'], label='2 Hidden Layers')
plt.plot(h3.history['val_accuracy'], label='3 Hidden Layers')
plt.plot(h4.history['val_accuracy'], label='4 Hidden Layers')

plt.title('So sánh Độ chính xác theo số lượng Lớp ẩn (Hidden Layers)-MNIST')
plt.xlabel('Epoch')
plt.ylabel('Validation Accuracy')
plt.legend()
plt.grid(True)
plt.show()
```

```python
lr_0_1     = train_model(num_hidden_layers=1, learning_rate=0.1)
lr_0_01    = train_model(num_hidden_layers=1, learning_rate=0.01)
lr_0_001   = train_model(num_hidden_layers=1, learning_rate=0.001)
lr_0_0001  = train_model(num_hidden_layers=1, learning_rate=0.0001)
plt.figure(figsize=(10, 6))

plt.plot(lr_0_1.history['loss'], label='LR = 0.1 ')
plt.plot(lr_0_01.history['loss'], label='LR = 0.01 ')
plt.plot(lr_0_001.history['loss'], label='LR = 0.001 ')
plt.plot(lr_0_0001.history['loss'], label='LR = 0.0001 ')

plt.title('MNIST - Learning Rate Comparison (Loss)')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)
plt.show()
```
